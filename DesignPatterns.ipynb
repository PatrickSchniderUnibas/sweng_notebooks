# Design Patterns: Beispiele

### Singleton

Der folgende Code soll einen Counter implementieren, der unabhängig von den kreierten Instanzen bei jeden Aufruf von `increase()` um 1 vergrössert ist. Die korrekte Ausgabe vom folgenden Beispiel müsste also

```
0

0

1

1

```
sein. Wie kannst du das erreichen?

````java
public class GlobalCounter {

    private int count;

    private GlobalCounter() {
        count = 0;
    }

    public static GlobalCounter create() {
        return new GlobalCounter();
    }

    public void increase() {
        count++;
    }

    public void print() {
        System.out.println(this.count);
    }

}
````

````java
GlobalCounter a = GlobalCounter.create();
GlobalCounter b = GlobalCounter.create();
a.print();
b.print();
a.increase();
a.print();
b.print();
````

### Abstract Factory

Ändere den folgenden Code so, dass stattdessen die Elemente für Windows ausgegeben werden. Wieviele Änderungen musst du dazu machen? Gibt es eine Möglichkeit, das Programm so umzuschreiben, dass du nur eine Änderung machen musst?

````java
// Abstract products
interface Button {
    void render();
}

interface Checkbox {
    void render();
}

// Concrete products
class WindowsButton implements Button {
    public void render() { System.out.println("Windows Button"); }
}

class MacButton implements Button {
    public void render() { System.out.println("Mac Button"); }
}

class WindowsCheckbox implements Checkbox {
    public void render() { System.out.println("Windows Checkbox"); }
}

class MacCheckbox implements Checkbox {
    public void render() { System.out.println("Mac Checkbox"); }
}

// Client
public class Main {
    public static void main(String[] args) {

        Button button = new MacButton();
        Checkbox checkbox = new MacCheckbox();

        button.render();    // Mac Button
        checkbox.render();  // Mac Checkbox
    }
}
````

````java
Main.main(new String[0]);
````
